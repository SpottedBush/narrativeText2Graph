```python
import torch
from gliner2 import GLiNER2

entities_to_extract = ["location", "fictionnal character", "actors"]

exctractor = GLiNER2.from_pretrained("fastino/gliner2-base-v1").to(torch.device("cuda" if torch.cuda.is_available() else "cpu"))
out = exctractor.extract_entities(
    text=doc.text,
    entity_types=entities_to_extract,
    threshold=0.5,
    include_confidence=True,
) # -> Dict[str, Any]
out
```

```python
def find_all_occurrences(text: str, pattern: str):
    """Return all start indexes where `pattern` appears in `text`."""
    indexes = []
    start = 0

    while True:
        idx = text.find(pattern, start)
        if idx == -1:
            break
        indexes.append((idx, idx + len(pattern)))
        start = idx + 1  # move forward to avoid infinite loops

    return indexes


all_occurences = []
for ent in out["entities"]["actors"]:
    occurences = find_all_occurrences(doc.text, ent)
    print(f"{occurences}, {ent} -> {doc.text[occurences[0][0]:occurences[0][1]]}")
    all_occurences.append(occurences)

for ent in out["entities"]["fictionnal character"]:
    occurences = find_all_occurrences(doc.text, ent)
    print(f"{occurences}, {ent} -> {doc.text[occurences[0][0]:occurences[0][1]]}")
    all_occurences.append(occurences)

for ent in out["entities"]["location"]:
    occurences = find_all_occurrences(doc.text, ent)
    print(f"{occurences}, {ent} -> {doc.text[occurences[0][0]:occurences[0][1]]}")
    all_occurences.append(occurences)

```

### GLINER

- **Joueurs (Players)** : Ce ne sont pas seulement des individus, mais aussi des entités impersonnelles qui prennent position, comme des agences gouvernementales, des organisations militantes ou des laboratoires scientifiques.
    
- **Positions** : Ce concept est plus large que celui de « point de vue » ou de « thèse ». Une position inclut les objectifs d'un joueur, ses préoccupations sous-jacentes, ses connaissances tacites et d'autres éléments qui constituent le matériau source de ses arguments.
    
- **Lieux (Places)** : Le concept le plus innovant, il désigne les écologies où se développe l'argumentation. Les lieux (une audience législative, un fil de discussion sur les réseaux sociaux, une revue scientifique) ne sont pas des contextes passifs ; ils influencent activement l'argumentation par leur design, leurs structures de participation et leurs normes de raisonnement.

```python
from gliner import GLiNER

# Initialize GLiNER with the base model
model = GLiNER.from_pretrained("urchade/gliner_medium-v2.1")
```

```python
from textwrap import wrap, fill

def display_entities_gliner(entities, original_text = None):
    if original_text:
        print(fill(original_text), "\n")

    for entity in entities:
        # print(entity)
        print(f'{entity["text"]:10s} => {entity["label"]:20s} | ({str(entity["start"]):4s},{str(entity["end"]):4s}) @ {entity["score"]:.2f}')
```

```python
entities_to_extract = ["location", "fictionnal character", "actors"]

# Perform entity prediction
entities = model.predict_entities(sample, entities_to_extract, threshold=0.5)
# entities = model.predict_entities(sample, polylogue_entities, threshold=0.5)

display_entities_gliner(entities, original_text=sample)
```
